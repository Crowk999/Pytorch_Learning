### Random Tesnor

### Random tensor is important beacuse our neural network start with random tensor then adjust those random number to provide the meaningful output.

```python
# random tensor of size (3,4) that also in minus
random_tensor = -torch.rand(3,4)
random_tensor
```

```python
random_image_size = torch.rand(size=(240,240,3)) # Height Width Channel
random_image_size.shape, random_image_size.ndim
```

```python
# Zero Tensors
zero_tensor = torch.zeros((1,2,3))
# One Tensors
one_tensors = torch.ones((1,2,3))
#print
one_tensors, zero_tensor
```

```python
# Creating range
one_to_ten = torch.arange(10,0,-5)
one_to_ten
```

```python
ten_zeros = torch.ones_like(one_to_ten)
ten_zeros
```

```python
data_type_32 = torch.tensor([1,2,3],
                            dtype=torch.int32,
                            device=None,
                            requires_grad=False)
data_type_32.dtype
```

```python
data_type_16 = data_type_32.type(torch.float16)
print(data_type_32*data_type_16)
```

DERIVATION

```python
#derivation
x = torch.tensor(2.0, requires_grad=True)
y = x**2
y.backward(x)
x.grad
```

GETTING INFORMATION FROM TENSOR

```python
some_tensor = torch.randint(0,10,(3,4), device="cpu", dtype=torch.int16)
# details of this tensor:
print("The tensor itself:-",some_tensor)
print("The datatype of tensor:-", some_tensor.dtype)
print("The size of tensor:-", some_tensor.shape) # or tensor.size()
print("The device tensor on:-", some_tensor.device)
print("The dimension of tensor", some_tensor.ndim)
```
