#Normal Arithmetic Operation    

```python
# Element wise operation
A = torch.randint(1,50,(2,3),dtype=torch.float16)
B = torch.randint(50,100,(3,))
sum = A.add(B) # torch.add(A,B)
diff = A.sub(B) # torch.sub(A,B)
mul = A.mul(B)
div = A.div(B)
print("ADD:",sum)
print("-----------------------------------")
print("SUB:",diff)
print("-----------------------------------")
print("MUL:",mul)
print("-----------------------------------")
print("DIV:",div)
```

```python
# Attributes one of those sum mul div and mean
alpha_sum = A.add(B, alpha=2) # Its A+2B
alpha_diff = A.sub(B, alpha=3) # Its A-3B
result = torch.empty(2,3) # This feels the matrix with garbage value
# A.mul(B, out = result) not allowed
torch.mul(A,B, out = result) # this out uses the same memory everytime so it saves meomory
# Also out can be used with all add sub div mul and matmul
rounding_div = A.div(B,rounding_mode="floor") # or trunc remove all the decimal but floor takes towards smaller one
# -5.1 for trunc is -5 but for floor its -6
print("A_Matrix:",A)
print("-----------------------------------")
print("B_Matrx:",B)
print("-----------------------------------")
print("ADD:",alpha_sum)
print("-----------------------------------")
print("MUl:",result)
print("-----------------------------------")
print("SUB:",alpha_diff)
print("-----------------------------------")
print("DIV:",rounding_div)
```

#Matrix Multiplication
1.    See the inner dimension of the both matrix must be same.
      i.e:-The  column of first and row of second must be same.
2.   The result matirx will have the shape of the row of first
      matrix and column of the second one. i.e:-A(B(x))



```python
# Pytorch wise
X = torch.tensor([
    [7,8],
    [5,6],
    [3,4],
    [1,2]
])# Its shape is (4,2)
Y = torch.randint(5,11,(2,10))
result = X@Y # result shape will be (4,10)
result # can also use A.matmul(B)
```

See matirx is a function which we use to see what happens to space if
we do something chnage on it like how to rotate flip and so on.

Thus Matrix Multiplication is A(B(x)). Thus we want to see the what will be chnage if we try to add the chnages sone by both matrix.   
i.e.- Output of B must match the input of A. Thus inner dimmension is same.

So here we multiply the row of with the column of B



```python
# Manual Matrix Multiplication loop
#row # col # elements
result = torch.zeros(4,10) # We assign Vlaue in Zeros
for i in range(len(X)): # Rows            -4
  for j in range(len(Y[0])): # Columns    -10
    for k in range(len(Y)): # Elements    -2 For X  its rows element for Y its columns elements
      result[i][j] += X[i][k]*Y[k][j]
result
```

# Resizing using reshape and squeezing and unsqueezing data
Reshape is used to resize the tensor


Squeezing means to remove the 1 dim and unsqueezing mean to add the one so dim 0 for first dim 1 for second and dim 3 for third so on however in squeezing only dim 1 works and dim 0 doesnt chnage and other dim gives error.




```python
# Resixing using reshape and squeezing and unsqueezing data
image=torch.arange(1,10,1).type(torch.int16)
print("Image:-",image)
print("-----------------------------------")
resized_image = image.reshape(3,1,3)
print("Resized Image:-",resized_image)
print("-----------------------------------")
sequeezed_image = resized_image.squeeze(dim=1)
print(f"Sequezed_image:-{sequeezed_image} and Its shape{sequeezed_image.shape}")
print("-----------------------------------")
unsequeezed_image = resized_image.unsqueeze(dim=1)
print(f"UnSequezed_image:-{unsequeezed_image} and Its shape{unsequeezed_image.shape}")

```
